for string type:

```python
def fillNaObjectMode(cols):
    for i in cols:
        df[i] = df[i].fillna(df[i].mode()[0])
        
columns = ['school', 'paid', 'activities']
fillNaObjectMode(columns)

```

for float type:

```python
def fillNaObjectMode(cols):
    for i in cols:
        df[i] = df[i].fillna(df[i].mean())
        
columns = ['Medu','Fedu','traveltime','studytime','failures','famrel',
    'freetime','goout','Dalc','Walc','health','absences','G1','G2','G3']
fillNaObjectMode(columns)
```

For int type:

```python
def fillNaObjectMode(cols):
    for i in cols:
        df[i] = df[i].fillna(df[i].mode()[0])
        
columns = ['Medu', 'Fedu', 'traveltime', 'studytime', 'failures',
    'famrel', 'freetime', 'goout', 'Dalc', 'Walc', 'health',
    'absences', 'G1', 'G2', 'G3']
fillNaObjectMode(columns)
```

## Manage data types ##

```python
df['studytime'] = df['studytime'].astype('int64')
```

```python
# For Float Type
def changeFloatToInt(cols):
    for i in cols:
        df[i] = df[i].astype('int64')

columns = ['Medu', 'Fedu', 'traveltime', 'studytime', 'failures', 'famrel',
'freetime', 'goout', 'Dalc', 'Walc', 'health', 'absences', 'G1', 'G2', 'G3'
]
changeFloatToInt(columns)
```

```python
def convertObjToInt(cols):
    for i in cols:
        data = pd.DataFrame({i:df[i].unique()})
        data_LE = LabelEncoder()
        data_LE.fit(np.ravel(data))    
        df[i] = data_LE.transform(df[i])
        
columns = ['school','sex','address','famsize','Pstatus','Mjob','Fjob','reason',
    'guardian','schoolsup','famsup','paid','activities','nursery','higher',
    'internet','romantic']
convertObjToInt(columns)
```
